```python
#2.11
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl

#khai báo đầu vào
dist = ctrl.Antecedent(np.arange(0,51,1), 'distance')
traffic = ctrl.Antecedent(np.arange(0,101,1), 'traffic')
demand = ctrl.Antecedent(np.arange(0,101,1), 'demand')
weather = ctrl.Antecedent(np.arange(0,11,1), 'weather')
rating = ctrl.Antecedent(np.arange(0, 5.1, 0.1), 'rating')
punc = ctrl.Antecedent(np.arange(0, 101, 1), 'punc')

#đầu ra
price = ctrl.Consequent(np.arange(0, 300000, 100), 'price')
bonus = ctrl.Consequent(np.arange(0, 101, 1), 'bonus')

#định mức
dist['short'] = fuzz.trimf(dist.universe, [0, 0, 3])
dist['med'] = fuzz.trimf(dist.universe, [0, 2, 10])
dist['long'] = fuzz.trimf(dist.universe, [6, 15, 20])
dist['vlong'] = fuzz.trimf(dist.universe, [15, 20, 50])

traffic['low'] = fuzz.trimf(traffic.universe, [0, 0, 30])
traffic['med'] = fuzz.trimf(traffic.universe, [20, 45, 70])
traffic['high'] = fuzz.trimf(traffic.universe, [60, 80, 100])

demand['low'] = fuzz.trimf(demand.universe, [0, 0, 30])
demand['med'] = fuzz.trimf(demand.universe, [20, 45, 70])
demand['high'] = fuzz.trimf(demand.universe, [60, 80, 100])

weather['bad'] = fuzz.trimf(weather.universe, [0, 0, 4])
weather['med'] = fuzz.trimf(weather.universe, [3, 5, 8])
weather['good'] = fuzz.trimf(weather.universe, [7, 9, 10])

rating['poor'] = fuzz.trimf(rating.universe, [1, 1, 2.5])
rating['med'] = fuzz.trimf(rating.universe, [2, 5, 7])
rating['good'] = fuzz.trimf(rating.universe, [6, 8, 10])

punc['late'] = fuzz.trimf(punc.universe, [0, 0, 50])
punc['ontime'] = fuzz.trimf(punc.universe, [40, 60, 80])
punc['soon'] = fuzz.trimf(punc.universe, [70, 90, 100])

price['low'] = fuzz.trimf(price.universe, [0, 20000, 50000])
price['med'] = fuzz.trimf(price.universe, [40000, 60000, 100000])
price['high'] = fuzz.trimf(price.universe, [90000, 120000, 150000])
price['vhigh'] = fuzz.trimf(price.universe, [130000, 200000, 300000])

bonus['none'] = fuzz.trimf(bonus.universe, [0, 0, 10])
bonus['few'] = fuzz.trimf(bonus.universe, [5, 15, 30])
bonus['med'] = fuzz.trimf(bonus.universe, [20, 45, 70])
bonus['high'] = fuzz.trimf(bonus.universe, [60, 80, 100])

#rules
rules = [
    #giá
    ctrl.Rule(dist['short'] & traffic['low'] & demand['low'], price['low']),
    ctrl.Rule(dist['short'] & traffic['med'] & demand['high'], price['med']),
    ctrl.Rule(dist['med'] & traffic['high'] & demand['high'], price['high']),
    ctrl.Rule(dist['long'] & traffic['med'] & weather['good'], price['med']),
    ctrl.Rule(dist['long'] & traffic['high'] & weather['bad'], price['vhigh']),
    ctrl.Rule(dist['vlong'] & traffic['high'] & weather['bad'], price['vhigh']),
    ctrl.Rule(dist['med'] & traffic['low'] & demand['low'], price['med']),
    ctrl.Rule(dist['short'] & traffic['high'] & weather['bad'], price['high']),
    ctrl.Rule(dist['vlong'] & weather['bad'], price['vhigh']),
    ctrl.Rule(dist['med'] & traffic['med'] & weather['med'], price['med']),
    ctrl.Rule(dist['long'] & traffic['high'] & weather['med'], price['high']),
    ctrl.Rule(demand['high'], price['high']),
    #bonus
    ctrl.Rule(rating['good'] & punc['soon'], bonus['high']),
    ctrl.Rule(rating['med'] & punc['ontime'], bonus['med']),
    ctrl.Rule(rating['poor'] & punc['late'], bonus['none']),
    ctrl.Rule(dist['long'] & traffic['high'] & punc['ontime'], bonus['high']),
    ctrl.Rule(dist['med'] & traffic['med'] & rating['good'], bonus['med']),
    ctrl.Rule(rating['poor'] & punc['late'], bonus['none']),
    ctrl.Rule(dist['vlong'] & weather['bad'] & rating['good'], bonus['high']),
    ctrl.Rule(dist['short'] & rating['med'] & punc['ontime'], bonus['few']),
    ctrl.Rule(dist['long'] & traffic['high'] & punc['late'], bonus['few']),
    ctrl.Rule(dist['med'] & weather['med'] & rating['good'], bonus['med']),
    ctrl.Rule(rating['med'] & punc['soon'], bonus['med'])

]

grab_sim = ctrl.ControlSystemSimulation(ctrl.ControlSystem(rules))
grab_sim.input['distance'] = 20
grab_sim.input['traffic'] = 80
grab_sim.input['demand'] = 90
grab_sim.input['weather'] = 8
grab_sim.input['rating'] = 4.8
grab_sim.input['punc'] = 95

grab_sim.compute()

print(f"Giá xe đề xuất: {grab_sim.output['price']} đồng")
print(f"Điểm thưởng tài xế: {grab_sim.output['bonus']:.2f}/100")
```

```python
#2.12
import numpy as np
import skfuzzy as fuzzy
from skfuzzy import control as ctrl

#đầu vào
rating = ctrl.Antecedent(np.arange(0, 5.1, 0.1), 'rating')
volume = ctrl.Antecedent(np.arange(0, 101, 1), 'volume')
margin = ctrl.Antecedent(np.arange(0, 101, 1), 'margin')
seasonal = ctrl.Antecedent(np.arange(0, 11, 1), 'seasonal')
competitor = ctrl.Antecedent(np.arange(0, 11, 1), 'competitor')

#đầu ra
discount = ctrl.Consequent(np.arange(0, 71, 1), 'discount')

#định mức
rating['low'] = fuzzy.trimf(rating.universe, [0, 0, 4])
rating['med'] = fuzzy.trimf(rating.universe, [3.8, 4.25, 4.5])
rating['high'] = fuzzy.trimf(rating.universe, [4.4, 5, 5])

volume['low']  = fuzzy.trimf(volume.universe, [0, 0, 50])
volume['med']  = fuzzy.trimf(volume.universe, [30, 50, 70])
volume['high'] = fuzzy.trimf(volume.universe, [60, 100, 100])

margin['low'] = fuzzy.trimf(margin.universe, [0, 0, 30])
margin['med'] = fuzzy.trimf(margin.universe, [20, 50, 70])
margin['high'] = fuzzy.trimf(margin.universe, [60,100, 100])

seasonal['none'] = fuzzy.trimf(seasonal.universe, [0, 0, 4])
seasonal['med'] = fuzzy.trimf(seasonal.universe, [3, 5, 7])
seasonal['high'] = fuzzy.trimf(seasonal.universe, [6, 10, 10])

competitor['low'] = fuzzy.trimf(competitor.universe, [0, 0, 4])
competitor['med'] = fuzzy.trimf(competitor.universe, [3, 5, 7])
competitor['high'] = fuzzy.trimf(competitor.universe, [6,10, 10])

discount['vlow'] = fuzzy.trimf(discount.universe, [0, 0, 5])
discount['low'] = fuzzy.trimf(discount.universe, [5, 7, 10])
discount['med'] = fuzzy.trimf(discount.universe, [10, 15, 20])
discount['high'] = fuzzy.trimf(discount.universe, [20, 30, 40])
discount['vhigh'] = fuzzy.trimf(discount.universe, [40, 70, 70])

rules = [
    ctrl.Rule(rating['high'] & volume['high'] & margin['high'], discount['vlow']),
    ctrl.Rule(rating['low'] & volume['low'] & margin['high'], discount['high']),
    ctrl.Rule(seasonal['high'] & competitor['high'], discount['vhigh']),
    ctrl.Rule(rating['med'] & volume['med'] & margin['med'], discount['med']),
    ctrl.Rule(competitor['low'] & margin['low'] & volume['high'], discount['vlow']),
    ctrl.Rule(rating['low'] & seasonal['none'], discount['med']),
    ctrl.Rule(volume['low'] & margin['low'], discount['high'])
]

shopee_sim = ctrl.ControlSystemSimulation(ctrl.ControlSystem(rules))

shopee_sim.input['rating'] = 4.3
shopee_sim.input['volume'] = 50
shopee_sim.input['margin'] = 15
shopee_sim.input['seasonal'] = 9
shopee_sim.input['competitor'] = 9

shopee_sim.compute()

print(f"Mức chiết khấu đề xuất: {shopee_sim.output['discount']:.2f}%")
```

```python
#2.14
import numpy as np
import skfuzzy as fuzzy
from skfuzzy import control as ctrl

#đầu vào
density = ctrl.Antecedent(np.arange(0, 11, 1), 'density')
urgency = ctrl.Antecedent(np.arange(0, 11, 1), 'urgency')
load = ctrl.Antecedent(np.arange(0, 11, 1), 'load')
traffic = ctrl.Antecedent(np.arange(0, 11, 1), 'traffic')
profit = ctrl.Antecedent(np.arange(0, 11, 1), 'profit')

#đầu ra
combine = ctrl.Consequent(np.arange(0, 11, 1), 'combine')
priority = ctrl.Consequent(np.arange(0, 101, 1), 'priority')

#định mức
density['low']  = fuzzy.trimf(density.universe, [0, 0, 4])
density['med']  = fuzzy.trimf(density.universe, [2, 5, 8])
density['high'] = fuzzy.trimf(density.universe, [6, 10, 10])

urgency['low']  = fuzzy.trimf(urgency.universe, [0, 0, 5])
urgency['med']  = fuzzy.trimf(urgency.universe, [3, 5, 7])
urgency['high'] = fuzzy.trimf(urgency.universe, [6, 10, 10])

load['low']  = fuzzy.trimf(load.universe, [0, 0, 3])
load['med']  = fuzzy.trimf(load.universe, [2, 4, 6])
load['high'] = fuzzy.trimf(load.universe, [5, 10, 10])

traffic['low']  = fuzzy.trimf(traffic.universe, [0, 0, 4])
traffic['med']  = fuzzy.trimf(traffic.universe, [3, 5, 7])
traffic['high'] = fuzzy.trimf(traffic.universe, [6, 10, 10])

profit['low']  = fuzzy.trimf(profit.universe, [0, 0, 4])
profit['med']  = fuzzy.trimf(profit.universe, [3, 5, 7])
profit['high'] = fuzzy.trimf(profit.universe, [6, 10, 10])

combine['few'] = fuzzy.trimf(combine.universe, [0, 0, 5])
combine['some'] = fuzzy.trimf(combine.universe, [3, 5, 7])
combine['many'] = fuzzy.trimf(combine.universe, [5, 10, 10])

priority['low'] = fuzzy.trimf(priority.universe, [0, 0, 50])
priority['med'] = fuzzy.trimf(priority.universe, [30, 50, 70])
priority['high'] = fuzzy.trimf(priority.universe, [50, 100, 100])

rules = [
    #luật kết hợp
    ctrl.Rule(density['high'] & load['low'] & traffic['low'], combine['many']),
    ctrl.Rule(density['med'] & traffic['high'] & urgency['med'], combine['some']),
    ctrl.Rule(load['high'] & density['high'] & profit['med'], combine['some']),
    ctrl.Rule(density['low'] & urgency['high'] & traffic['med'], combine['some']),
    ctrl.Rule(profit['high'] & urgency['high'] & traffic['high'], combine['some']),
    ctrl.Rule(urgency['med'] & traffic['med'] & profit['med'], combine['some']),
    #luật ưu tiên
    ctrl.Rule(urgency['high'] & profit['high'], priority['high']),
    ctrl.Rule(urgency['med'] & traffic['med'], priority['med']),
    ctrl.Rule(urgency['low'] & density['high'] & profit['low'], priority['low'])
]

logistics_ctrl = ctrl.ControlSystem(rules)
logistics_sim = ctrl.ControlSystemSimulation(logistics_ctrl)

logistics_sim.input['density'] = 9
logistics_sim.input['urgency'] = 5
logistics_sim.input['load'] = 2
logistics_sim.input['traffic'] = 5
logistics_sim.input['profit'] = 5

logistics_sim.compute()

print(f"Số lượng đơn hàng cần kết hợp: {logistics_sim.output['combine']:.1f} đơn")
print(f"Mức độ ưu tiên giao hàng: {logistics_sim.output['priority']:.1f}%")
```
